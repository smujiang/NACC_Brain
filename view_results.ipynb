```python
import pyvista as pv
import glob
import os

# directory containing vtk files
vtk_dir = "/Volumes/bai/NACC/MRI/test_run_output/temp/first_results"

# find all vtk files
vtk_files = sorted(glob.glob(os.path.join(vtk_dir, "*_first.vtk")))

print("Found files:")
for f in vtk_files:
    print(os.path.basename(f))

# create plotter
plotter = pv.Plotter()

# color palette

structure_colors = {
    "Hipp": "red",
    "Amyg": "yellow",
    "Caud": "blue",
    "Puta": "green",
    "Pall": "purple",
    "Thal": "cyan",
    "Accu": "orange",
    "BrStem": "brown"
}

plotter = pv.Plotter()

for vtk_file in vtk_files:
    
    mesh = pv.read(vtk_file)
    
    name = os.path.basename(vtk_file)
    
    color = "gray"
    for key in structure_colors:
        if key in name:
            color = structure_colors[key]
    
    plotter.add_mesh(mesh, color=color, smooth_shading=True)

plotter.add_legend([
    ("Hippocampus", "red"),
    ("Amygdala", "yellow"),
    ("Caudate", "blue"),
    ("Putamen", "green"),
    ("Pallidum", "purple"),
    ("Thalamus", "cyan"),
])

plotter.show()
```
